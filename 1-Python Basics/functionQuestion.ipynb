### Temperature Conversion

```python
#basically we use tem and its for convrting if its c the to f and f then to c

# function 
def convertTemp(temp,unit):
    if unit == 'C':
        return temp *9/5 + 32
    elif unit == 'F':
        return (temp-32)*5/9
    # else:
    #     return None
    # else function ni chlega kyuki woh function mai gusega hi ni compiler bcoz argumentss present ni hoge toh


print(convertTemp(25,'C')) 
print(convertTemp(90,'F'))
# print(convertTemp())
```

### Password Strength Checker

```python
# defining function in which we check if char is it in the form of digit , lower ,upper ,alpha or special character
def is_strong_pass(password):
    if len(password)<8:
        return False
    if not any(char.isdigit() for char in password):
        return False
    if not any(char.isupper() for char in password):
        return False
    if not any(char.islower() for char in password):
        return False
    # if not any(char.isalpha() for char in password):
    #     return False
    if not any(char in '@#$_' for char in password):
        return False
    return True

# using function to create a strong password
print(is_strong_pass('Weakpas'))
print(is_strong_pass('Ileen@14'))
```

### Calculate the Total Cost of Items in a Shooping Cart

```python
def total(cart):

  totalCost =0
  for item in cart:
     totalCost+= item['price'] * item['quantity']

  return totalCost


# Suppose this is the cart
cart =[
    {'name' : 'Pepsi' ,'price' : 25 ,'quantity' :5 },
    {'name' : 'CocaCola' ,'price' :30 ,'quantity' :10},
    {'name' :'Fanta' ,'price' : 10 ,'quantity' :15}
]
print(total(cart))
```

### Check if a String is Palindrome or not

```python
# so basically hmm isme ek pure sentence ko check krenge palindrome 
# Task - SPace remove krna hai , then lower case mai krr k pure string ko check for palindrome

def isPalindrome(str):
    # in python we can use two or more string methods at the same time
    str = str.lower().replace(" ","")
    # ab check krenge reverse is same or not
    return str == str[::-1]


word = "Never odd or even"
word2 ="November mai Job"
print(isPalindrome(word))
print(isPalindrome(word2))
```
