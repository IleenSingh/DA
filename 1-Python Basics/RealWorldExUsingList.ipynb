### Oraganising Student Grades
Create a list to store and calculate average grade of Students

```python
marks =[90 ,98 ,80 ,45,89]

# Calculating average grade 
avg = sum(marks) / len(marks)
print(f"Average Grade: {avg :.2f}")
# : .2f is used to indicate decimal literal

# Highest grade
h =max(marks)
# Lowest grade
l =min(marks)

print(f"Highest Grade : {h}")
print(f"Lowest Grade :{l}")
```

### Managing an Inventory

Use a lsit to manage inventory items in a store

```python
# Creating a fruit Inventory 
inventory =["Apple" , "Banana" ,"Cherry" ,"Watermelon"]
# removing watery fruit
inventory.remove("Watermelon")
# Checking for a specific element in a list
item = "orange"
if item in inventory:
    print(f"{item} is present in Inventory")
else :
    print(f"{item} is not present in Inventory")

print("Total items present in Fruit Inventory ")
for i in inventory :
    print(i)
```
