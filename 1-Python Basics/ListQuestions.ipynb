```python
# 1

positive_num = list(range(1,21))
print(positive_num)
```

```python
# 2

print(f"First Element -> {positive_num[0]}")
print(f"Middle Element -> {len(positive_num)//2}")
print(f"Last Element -> {positive_num[-1]}")
```

```python
print(f"first 5 elements -> {positive_num[:5]}")
print(f"last 5 elements -> {positive_num[-5:]}")
print(f"from 5to 15 elements -> {positive_num[5:16]}")

```
