### 5 main types of function arguments

### Positional Arguments

These are the most common type — the order of arguments matters.

```python
def add(a, b):
    print(a + b)

add(2, 3)
```

### Keyword Arguments

You specify the parameter name explicitly when calling the function.
Order does not matter.

```python
def student(name, age):
    print(name, age)

student(age=21, name="Riya")
```

### Default Arguments

You assign a default value to a parameter.
If no value is passed, the default one is used.

```python
def greet(name="Guest"):
    print("Hello", name)

greet()         # Uses default → Guest
greet("Ileen")   # Overrides default
```

### Variable-Length Arguments (*args)

Used when you don’t know how many positional arguments will be passed.

```python
def total(*numbers):
    print(sum(numbers))

total(2, 3, 4)
```

### Keyword Variable-Length Arguments (**kwargs)

Used when you don’t know how many keyword arguments will be passed.

```python
def info(**data):
    print(data)

info(name="Ileen", age=22, city="Noida")
```
