### dict delete methods


```python
# del: Removes an item by key.
# pop(): Removes an item by key and returns its value.
# clear(): Empties the dictionary.
# popitem(): Removes and returns the last key-value pair.

d = {1: "Ileen" , 2: "Singh" ,3:"Frooti" ,4:"Singh"}
del d[4]
print(d)

```

```python
val = d.pop(3)
print(val)
```

```python
key ,value =d.popitem()
print(key,value)
```

```python
d.clear()
print(d)
```
