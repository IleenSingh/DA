### filter() Function

The filter() function in Python selects elements from an iterable for which a given function returns True.

In simple words:
👉 It filters data based on a condition

```python
def isEven(x):
    if x%2==0:
        return True


elements = [12,33,23,44,55,65,79,100]
answer = list(filter(isEven,elements))
print(answer)
```
